# LLM on UNOQ: GGUF + llama.cpp

This notebook is for **running** small LLMs on Arduino UNOQ (Linux side) using `llama.cpp` and a quantized **GGUF** model.

Practical note: UNOQ is for inference. If you need fine-tuning, do it on a PC and then convert/quantize to GGUF.


## Environment (UNOQ)

- Python: `3.11.x` recommended
- Build tools for `llama.cpp`: `build-essential`, `cmake`, `git`

Start Jupyter on UNOQ:

```bash
cd ~/qa-intermediate-curriculum/projects/LLM
python -m venv .venv
source .venv/bin/activate
pip install notebook
python -m notebook --ip 0.0.0.0 --port 8888 --no-browser
```


## Build llama.cpp (once)

```bash
sudo apt-get install -y build-essential cmake git
git clone https://github.com/ggerganov/llama.cpp
cd llama.cpp
cmake -S . -B build
cmake --build build -j
```


## Put a GGUF model on UNOQ

Place the model under (example):

- `~/qa-intermediate-curriculum/projects/LLM/models/<model>.gguf`

Recommended starting points (GGUF, Q4):

- TinyLlama 1.1B (Q4)
- Qwen2/Qwen2.5 0.5B (Q4)


````python
from pathlib import Path

LLAMA_CPP_DIR = Path('~/llama.cpp').expanduser()
MODEL = Path('models/gemma-4-12b-it-qat-q4_0.gguf')

LLAMA_CLI = LLAMA_CPP_DIR / 'build' / 'bin' / 'llama-cli'
LLAMA_CLI, MODEL
````

````python
import shlex
import subprocess

prompt = "You are a helpful assistant. In one sentence, explain what a wake word is."

cmd = [
    str(LLAMA_CLI),
    '-m', str(MODEL),
    '-c', '512',
    '-n', '128',
    '--temp', '0.7',
    '--top-p', '0.95',
    '--single-turn',
    '-p', prompt,
]

print('Running:', ' '.join(shlex.quote(x) for x in cmd))
subprocess.check_call(cmd)
````
